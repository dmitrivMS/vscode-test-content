# Data Analysis: Sales Trends
Exploring quarterly sales data with pandas and matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = {
    'quarter': ['Q1', 'Q2', 'Q3', 'Q4'],
    'revenue': [42000, 51000, 47500, 63000],
    'expenses': [38000, 42000, 40000, 45000]
}
df = pd.DataFrame(data)
df['profit'] = df['revenue'] - df['expenses']
df.head()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(df['quarter'], df['revenue'], label='Revenue', alpha=0.8)
ax.bar(df['quarter'], df['expenses'], label='Expenses', alpha=0.8)
ax.set_ylabel('Amount ($)')
ax.set_title('Quarterly Financial Overview')
ax.legend()
plt.tight_layout()
plt.show()
```
